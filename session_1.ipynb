```python
# Question - 1: how can you convert celsius to fahrenheit and produce output that proves each required step was performed?

celsius = 0.0

scaled = celsius * 9 / 5
fahrenheit = scaled + 32

print("Initial temperature:", celsius, "C")
print("Temperature in Fahrenheit:", fahrenheit, "F")
```

```python
# Question - 5: how can you convert total seconds to hours, minutes, and seconds and produce output that proves each required step was performed?

total_seconds = 7384

hours = total_seconds // 3600

remaining_after_hours = total_seconds % 3600

minutes = remaining_after_hours // 60

seconds = remaining_after_hours % 60

print("hours :", hours , "minutes :", minutes ,"secomnds :", seconds)
```

```python
# Question - 6: how can you calculate a bill after percentage discount and produce output that proves each required step was performed?

price = 2400.0
discount_rate =  0.15

discount_amount = price * discount_rate
final_price = price - discount_amount

print("final price:" , final_price)
```

```python
# Question - 7: how can you swap two inputs without a temporary variable and produce output that proves each required step was performed?

first_value = 12
second_value = 45

original_first = first_value
original_second = second_value

print("before swap -> first_value:", first_value, "second_value:", second_value)

first_value, second_value = second_value, first_value


print("after swap -> first_value:" , first_value, "second_value:", second_value)
print("correctness check -> first_value == original_second:", first_value == original_second)
print("correctness check -> second_value == original_first:", second_value == original_first)

```
