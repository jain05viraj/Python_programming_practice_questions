```python
# Question - 1: how can you compute quotient and remainder and produce output that proves each required step was performed?

dividend = 137
divisor = 60

quotient = dividend // divisor
remainder = dividend % divisor

verification = quotient * divisor + remainder

print("Dividend:", dividend ,"Divisor:", divisor ,"Quotient:", quotient ,"Remainder:", remainder)
print("Verification = Quotient * Divisor + Remainder =", verification)
```

```python
# Question - 2: how can you round a measurement to two places and produce output that proves each required step was performed?

measurement = 18.3764

round_measurement = round(measurement , 2)

print("Rounded measurement =", round_measurement)
```

```python
# Question - 3: how can you convert rupees to paise exactly and produce output that proves each required step was performed?

rupees = 125.75

paise_intermediate = rupees * 100
paise = round(paise_intermediate)

print("Rupees :", rupees)
print("Paise :", paise)
```

```python
# Question - 4: how can you calculate compound growth and produce output that proves each required step was performed?

principal = 50000.0
annual_rate = 0.08
years = 2

growth_multiplier = (1 + annual_rate) ** years
final_amount = principal * growth_multiplier
growth_amount = final_amount - principal

print("Principal amount:",principal)
print("Final amount:",final_amount)
print("Growth amount",growth_amount)
```

```python
# Question - 5:  how can you find distance between two numeric points and produce output that proves each required step was performed?

point_a = 42.75
point_b = 18.5

raw_difference = point_b - point_a

distance = abs(raw_difference)
print("initial points = point a:",point_a,"point b:",point_b)
print("Distance = ",distance)
```

```python
# Question - 6: how can you test divisibility by 3 and 5 and produce output that proves each required step was performed?

number = 45

remainder_by_3 = number % 3
remainder_by_5 = number % 5

divisible_by_3 = remainder_by_3 == 0
divisible_by_5 = remainder_by_5 == 0

print("Number =",number)
print("Remainder when divided by 3 =",remainder_by_3)
print("Remainder when divided by 5 =",remainder_by_5)
print("Final result = divisible by 3:",divisible_by_3 ,"divisible by 5:", divisible_by_5)
```

```python
# Question - 7: how can you convert bytes to kb and mb and produce output that proves each required step was performed?

byte_count = 5242880

kilobytes = byte_count / 1024

megabytes = kilobytes / 1024

print("Bytes =", byte_count)
print("Kilobytes =", kilobytes)
print("Megabytes =", megabytes)
```

```python
# Question - 8: how can you evaluate a polynomial for x and produce output that proves each required step was performed?

x = 4
a = 2
b = 3
c = 5

squared_term = a * x ** 2
linear_term = b * x
polynomial_value = squared_term + linear_term + c

print("input values -> x =",x ,"a =",a ,"b =",b ,"c =", c)
print("Squared term =", squared_term)
print("linear term =", linear_term)
print("Final output =", polynomial_value)
```
