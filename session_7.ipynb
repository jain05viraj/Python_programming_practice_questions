```python
# Question 1 : how can you use extended sequence unpacking and produce output that proves each required step was performed?

values = [10, 20, 30, 40, 50]

first, *middle, last = values

print("First Value =", first)
print("Middle value =", middle)
print("Last Value =", last)
```

```python
# Question 2 : how can you demonstrate equality versus identity and produce output that proves each required step was performed?

first = [1, 2]
second = [1, 2]

alias = first

equality_check = first == second
identity_check_second = first is second
identity_check_alias = first is alias

print(" Starting values : First value:", first ,"Second value :",second)
print("Equality check : First == Second ", equality_check)
print(" First is Second", identity_check_second)
print(" First is Alias", identity_check_alias)
```

```python
# Question 3 : how can you swap three variables by unpacking and produce output that proves each required step was performed?

a = 'North'
b = 'South'
c = 'East'

print("Before rotation : a:", a, "b:", b,"c:", c)

a, b, c = c, a, b

print("After rotation : a:", a, "b:", b, "c:", c)
```

```python
# Question 4 : how can you avoid chained-assignment aliasing and produce output that proves each required step was performed?

first = []
second = []

same_object = first is second

print("Starting values = First:", first ,"Second:", second)
print("First is Second :", same_object)

first.append(1)

print("Appended values = First:", first, "Second:", second)

```

```python
# Question 5 : how can you update a list with augmented assignment and produce output that proves each required step was performed?

values = [10, 20]
extra_values = [30, 40]

print("Starting values = value:", values ,"Extra value:", extra_values)

id_before = id(values)

values += extra_values

id_after = id(values)

print("final values :", values)

same_object = id_before == id_after

print("ID Is same :", same_object)
```

```python
# Question 6 : how can you capture a computed value with the walrus operator and produce output that proves each required step was performed?

number = [3, 8, 12]

print("Starting numbers:", number)
print("Highest number:", (Highest := max(number)))

greater_than_10 = Highest > 10

print("Highest number is > 10: ", greater_than_10)
```
