```python
# Question - 1: how can you create, index, and slice a product list and produce output that proves each required step was performed?

products = ['Pen', 'Notebook', 'Marker', 'Folder', 'Stapler']

first_product = products[0]
last_product = products[-1]
middle_slice = products[1:4]
reversed_products = products[::-1]

print("list =", products)
print("First product =", first_product)
print("Last product =", last_product)
print("Middle slice =", middle_slice)
print("Reverse =", reversed_products)
```

```python
# Question - 2: how can you compare append() with extend() and produce output that proves each required step was performed?

items = ['Pen', 'Notebook']
new_items = ['Marker', 'Folder']

items_for_append = items.copy()
items_for_extend = items.copy()

items_for_append.append(new_items)
items_for_extend.extend(new_items)

appended_result = items_for_append
extended_result = items_for_extend

print("Items:", items, "New items:", new_items)
print("appended list:", appended_result)
print("extended list:", extended_result)

```

```python
# Question - 3: how can you use insert(), remove(), and pop() and produce output that proves each required step was performed?

cities = ['Delhi', 'Mumbai', 'Pune', 'Jaipur']
removed_city = 'Pune'

print("cities :", cities)
cities.insert(1, 'Chennai')
before_remove = cities.copy()
print(" Cities after insert()", cities)

cities.remove(removed_city)
after_remove = cities.copy()
print("cities after remove:", after_remove)

popped_city = cities.pop()
print("after pop:", popped_city)
```

```python
# Question - 4: how can you compare sort() with sorted() and produce output that proves each required step was performed?

scores = [72, 91, 68, 85, 77]
print("Initial value", scores)

sorted_score = sorted(scores)
print("Sorted scores:", sorted_score)

scores.sort(reverse=True)
print("sort scores:", scores)
```

```python
# Question - 5:how can you retrieve values from a nested list and produce output that proves each required step was performed?

Matrix = [[10, 20], [30, 40], [50, 60]]

row_1 = Matrix[0]
row_2 = Matrix[1]
row_3 = Matrix[2]

item_1 = row_1[0]
item_2 = row_2[1]
item_3 = row_3[1]

print(" initial value =", Matrix)
print("row selection =", row_1 ,"| item selection =", item_1)
print("row selection =", row_2 ,"| item selection =", item_2)
print("row selection =", row_3 ,"| item selection =", item_3)
```

```python
# Question - 6: how can you demonstrate list aliasing and produce output that proves each required step was performed?

original = [10, 20, 30]
alias = original
same_object = alias is original 
print("initial value", original)
print("Alias :", alias ,"Same object =", same_object)
```

```python
# Question - 7: how can you make an explicit independent nested copy and produce output that proves each required step was performed?

original = [[1], [2]]

independent = [original[0].copy(), original[1].copy()]

is_independent = independent[0] is not original[0]
print("Identity check :", is_independent)
original[0].append(99)
print("Before append:", original)
print("after append:", independent)
```

```python
# Question - 8: how can you combine and repeat lists and produce output that proves each required step was performed?

first_list = ['A', 'B']
second_list = ['C', 'D']
repeat_count = 2

combined = first_list + second_list
repeated = combined * repeat_count

print("first list:", first_list ,"second list:", second_list)
print("Combimned list:", combined ," Repeated list:", repeated)

```
