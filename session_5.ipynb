```python
# Question - 1: how can you create and retrieve a product record and produce output that proves each required step was performed?

product = {'id': 101, 'name': 'Notebook', 'price': 125.0}

product_id = product['id']
product_name = product['name']
product_price = product['price']

print("Product =", product)
print("Product id =", product_id)
print("Product name =", product_name)
print("Product price =", product_price)
```

```python
# Question - 2:  how can you add and update dictionary entries and produce output that proves each required step was performed?

product = {'name': 'pen', 'price': 20.0}
stock = 50 
revised_price = 22.0
print("initial product:", product)

product_before = product.copy()
product['stock'] = stock
product['price'] = revised_price
print("final result:", product)
```

```python
# Question - 3: how can you use get() for a possibly missing key and produce output that proves each required step was performed?

employee = {'id': 'E101', 'name': 'Ravi'}
requested_key = 'department'

key_present_before = requested_key in employee

print("initially:", employee)
print("key present:", key_present_before)

department_value = employee.get(requested_key, 'Not available')

print("department value:", department_value)
```

```python
# Question - 4: how can you remove an entry with pop() and produce output that proves each required step was performed?

cart = {'Pen': 3, 'Notebook': 3, 'Marker': 1}
item_to_remove = 'Marker'

print("initial cart :", cart)
print("item to be removed:", item_to_remove)

removed_quantity = cart.pop(item_to_remove)
print("Removed quantity =", removed_quantity)
print("Remaining cart:", cart)
```

```python
# Question - 5: how can you retrieve values from a nested dictionary and produce output that proves each required step was performed?

student = {'name': 'Asha', 'marks': {'Python': 82, 'Statistics': 76}}
marks_dict = student['marks']

python_marks = marks_dict['Python']
statistics_marks = marks_dict['Statistics']
print("starting information:", student)
print("Python markes:", python_marks ," Statistics marks:", statistics_marks)
```

```python
# Question - 6: how can you merge two dictionaries with the union operator and produce output that proves each required step was performed?

defaults = {'theme': 'light', 'page size': 20}
user = {'theme': 'dark'}

settings = defaults | user
print("Merged dictionries =", settings)
```

```python
# Question - 7: how can you inspect keys, values, and items views and produce output that proves each required step was performed?

inventory = {'Pen': 40 , 'Notebook': 25, 'Marker': 18}
key_view = inventory.keys()
values_view = inventory.values()
items_view = inventory.items()

print("Inventory:", inventory)
print("Final keys view:", key_view)
print("Final value view:", values_view)
print("Final item view:", items_view)
```
