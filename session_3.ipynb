```python
# Question - 1: how can you count vowels case-insensitively and produce output that proves each required step was performed?

text = 'Data Science'

lower_text = text.lower()

count_a = lower_text.count('a')
count_e = lower_text.count('e')
count_i = lower_text.count('i')
count_o = lower_text.count('o')
count_u = lower_text.count('u')

print("Initial input text =", text)
print("Pre vowel counts -> a:", count_a ,"e:", count_e ,"i:", count_i ,"o:", count_o ,"u:", count_u)

vowel_count = count_a + count_e + count_i + count_o + count_u

print("Total vowels =", vowel_count)
```

```python
# Question - 2: how can you reverse text without reversed and produce output that proves each required step was performed?

text = 'Python'

reversed_text = text[::-1]

print("Initial text =",text)
print("Final text =",reversed_text)
```

```python
# Question - 3: how can you test a normalized palindrome and produce output that proves each required step was performed?

text = '  Never Odd Or Even  '

normalized_text = text.strip().lower().replace(' ', '')
reversed_normalised = normalized_text[::-1]

is_palandrome = normalized_text == reversed_normalised
print("The given input is:", text)
print("and the given text is a palandrome:", is_palandrome)

```

```python
# Question - 4: how can you extract a filename extension and produce output that proves each required step was performed?

filename = 'archive.tar.gz'

name_pieces = filename.rsplit('.', 1)
extension = name_pieces[1]

print("filename:",filename)
print("extension:",extension)

```

```python
# Question - 5: how can you replace repeated whitespace and produce output that proves each required step was performed?

text = 'Python   makes   text   readable'

triple_space = '   '
single_space = ' '

cleaned_text = text.replace(triple_space, single_space)

print("starting text:", text)
print("final text:", cleaned_text)
```

```python
# Question - 6: how can you create initials from a full name and produce output that proves each required step was performed?

full_name = 'Asha Mehta Singh'

name_parts = full_name.split()

first_initial = name_parts[0][0]
second_initial = name_parts[1][0]
third_initial = name_parts[2][0]

initials = first_initial + '.' + second_initial + '.' + third_initial 

print("Full name:", full_name)
print("Initials:", initials)
```

```python
# Question - 7: how can you count occurrences of a word and produce output that proves each required step was performed?

sentence = 'data helps data teams'
target = 'data'

word_count = sentence.count(target)

print("sentence :", sentence, "Target word :", target)
print("Word count =", word_count)
```
